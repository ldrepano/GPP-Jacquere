```python
#import data with the use of tracrRNA v2 (for fair comparison to Jacquere screen)
reads_df = pd.read_excel("../Data/Brunello_Sanson2018_SuppData1.xlsx",sheet_name = 'A375_mod_tracr raw reads', skiprows=1)
reads_df
```

```python
#read in annotations
annotations_df = pd.read_csv("../Data/brunello_ensemblv113_guidemappings.csv")
annotations_df['On-target Gene Symbols'] = annotations_df['On-target Gene Symbols'].fillna(annotations_df['Notes'])
annotations_df_melted = pd.concat([annotations_df['Target Sequence'],annotations_df['On-target Gene Symbols'].str.split(',',expand=True)],
                                  axis=1).melt(id_vars = 'Target Sequence', value_name = 'Gene Symbol').dropna().drop('variable',axis=1)
annotations_df = annotations_df_melted.drop_duplicates()
annotations_df = annotations_df.rename(columns = {'Target Sequence':'sgRNA Sequence'})
```

### Z-Score

```python
def lognorm(reads):
    """
    Standardize read counts by calculating reads per million,
    adding a pseudo-count of one, and taking the log2

    :param reads: numpy or pandas array
    :returns: numpy or pandas array
    """
    reads_per_million = (reads/reads.sum())*(10**6)
    lognormed_reads = np.log2(reads_per_million + 1)
    return lognormed_reads
```

```python
condition_columns = ['pDNA','RepA','RepB','RepC']
lognorm_df = reads_df.copy()
for column in condition_columns:
    lognorm_df[column] = lognorm_df[column].astype(float)
lognorm_df[condition_columns] = lognorm_df[condition_columns].apply(lognorm)
lognorm_df
```

```python
filtered_lognorms = lognorm_df.copy()
pdna_cols = ['pDNA']
z_low = -3  # minimum z-score
# Z-score the pDNA columns
z_scored_cols = []
for pdna in pdna_cols:
    z_col = pdna + '_z'
    filtered_lognorms[z_col] = (filtered_lognorms[pdna] - filtered_lognorms[pdna].mean())/filtered_lognorms[pdna].std()
    z_scored_cols.append(z_col)
# Filter by z-score
filtered_lognorms = filtered_lognorms[filtered_lognorms[z_scored_cols].min(axis = 1) > z_low]
# Drop z-scored columns
filtered_lognorms = filtered_lognorms.drop(z_scored_cols, axis=1)
filtered_lognorms
```

```python
print('Filtered ' + str(lognorm_df.shape[0] - filtered_lognorms.shape[0]) + ' sgRNAs with low pDNA')
```

```python
lfc_df = filtered_lognorms.copy()
target_cols = reads_df.columns[reads_df.columns.str.contains('Rep')].tolist()
for target_col in target_cols:
    # use lognorm df in case columns double as target and ref
    lfc_df[target_col] = filtered_lognorms[target_col] - filtered_lognorms[pdna_cols[0]]

# Remove reference column
lfc_df = lfc_df.drop(pdna_cols, axis=1)

lfc_df
```

```python
lfc_correlations = lfc_df[target_cols].corr()
sns.clustermap(lfc_correlations, cmap='RdBu_r',
               vmin=-1, vmax=1, annot=True)
```

```python
long_lfcs = (lfc_df[target_cols + ['sgRNA Sequence']].melt(id_vars='sgRNA Sequence',var_name='condition', value_name='lfc'))
avg_lfcs = (long_lfcs.groupby(['sgRNA Sequence'])
            .agg(avg_lfc = ('lfc', 'mean'),
                 n_obs = ('lfc', 'count'))
            .reset_index())
annotated_sgrna_lfcs = (avg_lfcs.merge(annotations_df,
                                       how='inner',
                                       on='sgRNA Sequence'))
annotated_sgrna_lfcs
```

```python
control_sgrna_lfcs = (annotated_sgrna_lfcs[(annotated_sgrna_lfcs['Gene Symbol']
                                            .str.contains('ONE_INTERGENIC'))]
                      .reset_index(drop=True))
control_sgrna_lfcs
```

```python
neg_ctl_mean = np.mean(control_sgrna_lfcs['avg_lfc'])
neg_ctl_sd = np.std(control_sgrna_lfcs['avg_lfc'])
zscored_sgrna_lfcs = annotated_sgrna_lfcs.copy()
zscored_sgrna_lfcs['z_scored_avg_lfc'] = ((zscored_sgrna_lfcs['avg_lfc'] - neg_ctl_mean)/neg_ctl_sd)

zscored_sgrna_lfcs
```

```python
num_pseudogene = 1000
not_control_sgrna_lfcs = (zscored_sgrna_lfcs.loc[(~zscored_sgrna_lfcs['Gene Symbol']
                                            .str.contains('ONE_INTERGENIC')) &
                                              (~zscored_sgrna_lfcs['Gene Symbol']
                                            .str.contains('NO_SITE')) &
                                              (~zscored_sgrna_lfcs['Gene Symbol']
                                            .str.contains('MULTIPLE_INTERGENIC')),:]
                      .reset_index(drop=True))
n_guides = not_control_sgrna_lfcs.groupby('Gene Symbol')['sgRNA Sequence'].nunique().mode()[0]
control_sgrnas = control_sgrna_lfcs['sgRNA Sequence'].unique()
pseudogene_df_list = []

random.seed = 0
for i in range(num_pseudogene):
     if i%100 == 0:
         print(i)
     chosen_guides = random.choices(control_sgrnas, k=n_guides)
     temp_df_list = []
     for j in range(n_guides):
         g = chosen_guides[j]
         temp_row = zscored_sgrna_lfcs.loc[(zscored_sgrna_lfcs['sgRNA Sequence'] == g) &
                                           (zscored_sgrna_lfcs['Gene Symbol'].str.contains('ONE_INTERGENIC')),:]
         temp_row.loc[:,'sgRNA Sequence'] = g + 'p' + str(i) + 'g' + str(j)
         temp_df_list.append(temp_row)
     temp_df = pd.concat(temp_df_list)
     temp_df.loc[:,'Gene Symbol'] = 'Pseudogene_' + str(i)
     pseudogene_df_list.append(temp_df)

pseudogene_df = pd.concat(pseudogene_df_list).reset_index(drop=True)
pseudogene_df
```

```python
n_guides
```

```python
zscored_sgrna_lfcs = pd.concat([not_control_sgrna_lfcs,pseudogene_df]).reset_index(drop=True)
#save sgrna level values
zscored_sgrna_lfcs.to_csv('../Data/files_for_hitcalling_methods/Z-Score/zscored_sgrna_lfcs_v3.txt', sep = '\t', index = False)
```

```python
zscored_gene_lfcs = (zscored_sgrna_lfcs.groupby(['Gene Symbol'])
                     .agg(z_scored_lfc_sum = ('z_scored_avg_lfc', 'sum'),
                          n_sgrnas = ('avg_lfc', 'count'))
                     .reset_index())
zscored_gene_lfcs['z_scored_avg_lfc'] = (zscored_gene_lfcs['z_scored_lfc_sum']/np.sqrt(zscored_gene_lfcs['n_sgrnas']))
zscored_gene_lfcs = zscored_gene_lfcs.drop('z_scored_lfc_sum', axis=1)
zscored_gene_lfcs = zscored_gene_lfcs[zscored_gene_lfcs['n_sgrnas'] > 1]
zscored_gene_lfcs = zscored_gene_lfcs[~(zscored_gene_lfcs['Gene Symbol'] == '(ENSG000')].reset_index(drop=True)
zscored_gene_lfcs['p_value'] = (zscored_gene_lfcs['z_scored_avg_lfc']
                                .transform(lambda x: scipy.stats.norm.sf(abs(x))*2))
zscored_gene_lfcs
```

```python
#save gene level values
zscored_gene_lfcs.to_csv('../Data/files_for_hitcalling_methods/Z-Score/zscored_gene_lfcs_v3.txt', sep = '\t', index = False)
```

### Determine STARS Threshold Parameter
Use the relative ranking of negative controls among all guides in the library to approximate the fraction of guides that "are receiving a meaningful rank" (the assumption here is that the ranks of negative controls are random/meaningless).

```python
annotated_sgrna_lfcs = annotated_sgrna_lfcs.sort_values(by = 'avg_lfc').reset_index(drop=True)
annotated_sgrna_lfcs['rank'] = annotated_sgrna_lfcs.index+1
```

```python
control_sgrnas = annotated_sgrna_lfcs.loc[(annotated_sgrna_lfcs['Gene Symbol'].str.contains('ONE_INTERGENIC')),
                                          'sgRNA Sequence'].drop_duplicates()
num_controls = len(control_sgrnas)

cutoffs = np.linspace(0.01,1,101)
fdr_vals = []
for cutoff in cutoffs:
    rank = np.floor(cutoff*len(annotated_sgrna_lfcs))
    above_cutoff = annotated_sgrna_lfcs.loc[annotated_sgrna_lfcs['rank'] <= rank, 'sgRNA Sequence']
    fdr = sum(above_cutoff.isin(control_sgrnas))/num_controls
    fdr_vals.append(fdr)
plt.plot(cutoffs,fdr_vals)
cutoff = max(cutoffs[[x <= 0.05 for x in fdr_vals]])
cutoff
```

The top 16% of guides (when ranked by LFC) include 5% of controls. 16% is therefore an appropriate value for the STARS threshold parameter. Note that 5% is a completely arbitrary threshold.

### MAGeCK input file
The MAGeCK input file must include raw read counts for each construct along with its corresponding target. Control guide RNAs are not included in MAGeCK output so it is necessary to duplicate them in the input file so that FDR can be calculated later on.

```python
final_guide_gene_annotation = zscored_sgrna_lfcs[['sgRNA Sequence','Gene Symbol']].drop_duplicates()

#must merge on the guide sequence without pseudogene identifer appended because that is what is present in reads_df
final_guide_gene_annotation.loc[:,'sgRNA_to_merge_on'] = final_guide_gene_annotation['sgRNA Sequence'].str.split('p',expand=True)[0]
annotated_reads_df = final_guide_gene_annotation.merge(reads_df.rename(columns = {'sgRNA Sequence':'sgRNA_to_merge_on'}),
                                                                       how = 'left',
                                                                       on = 'sgRNA_to_merge_on')
#dropping A549 columns since we are only interested in analyzing A375 data for now
annotated_reads_df = annotated_reads_df.drop(['sgRNA_to_merge_on'],axis=1)

#creating a copy of all guides chosen to target pseudogenes
pseudogene_reads_df = annotated_reads_df[annotated_reads_df['Gene Symbol'].str.contains('Pseudogene')]
pseudogene_reads_df.loc[:,'Gene Symbol'] = pseudogene_reads_df.loc[:,'Gene Symbol'] + '_copy'
annotated_reads_df = pd.concat([annotated_reads_df,pseudogene_reads_df],axis=0)

#MAGeCK throws an error when the same guide targets multiple genes, so make sgRNA Sequence unique by appending gene symbol
annotated_reads_df['sgRNA Sequence'] = annotated_reads_df['sgRNA Sequence'] + annotated_reads_df['Gene Symbol']
annotated_reads_df

annotated_reads_df.to_csv('../Data/files_for_hitcalling_methods/Mageck/brunello_counts_v3.txt', sep='\t', index=False)

control_sgrnas = annotated_reads_df.loc[annotated_reads_df['Gene Symbol'].str.contains('copy'),'sgRNA Sequence']

control_sgrnas.to_csv('../Data/files_for_hitcalling_methods/Mageck/brunello_controls_v3.txt', index=False, header = False)
```
