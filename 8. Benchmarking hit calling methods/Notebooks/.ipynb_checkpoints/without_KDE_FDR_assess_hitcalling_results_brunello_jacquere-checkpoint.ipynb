Comparing the results of various hit-calling methods on Brunello, Jacquere screening data. The Z-score method is calculated in the Jacquere Screen Performance section, and hit-calling with the hypergeometric and MAGeCK methods are detailed in the run_hitcallingmethods notebooks in this section. The hypergeometric and MAGeCK files read below are the direct outputs of running these methods.


```python
zscore_withFDR= pd.read_csv("../../6. Jacquere Screen Performance/Data/jacquere_brunello_zscoremethod_withFDR.csv")
zscore_withFDR['p_value'] = (zscore_withFDR['gene_zscore'].transform(lambda x: scipy.stats.norm.sf(abs(x))*2))

jacquere_zscore_results=zscore_withFDR[zscore_withFDR["condition"]=="Jacquere A375"]
jacquere_hypergeometric_results=pd.read_table("../Data/jacquere_hypergeometric_results.txt")
jacquere_mageck_results=pd.read_table("../Data/jacquere_A375_mageck_results.txt")

brunello_zscore_results=zscore_withFDR[zscore_withFDR["condition"]=="Brunello A375"]
brunello_hypergeometric_results=pd.read_table("../Data/brunello_hypergeometric_results.txt")
brunello_mageck_results=pd.read_table("../Data/brunello_A375_mageck_results.txt") 
```

```python
def process_hitcalling_results(zscore_df,hypergeometric_df,mageck_df):
    '''
    Inputs: Dataframes with results of the z-score method, hypergeometric (), and MAGeCK on the same negative selection CRISPR screen 
    Harmonizes formatting and calculates False Discovery Rate with pseudogenes from negative controls (in input to hit-calling methods)
    Output: Dataframe that concatenates the results of all three methods for compatibility with plots, downstream analysis 
    '''

    #arbitrarily picked one dataframe to retrieve the number of Pseudogenes; should be the same for all 
    num_pseudogene = hypergeometric_df.loc[hypergeometric_df['Gene Symbol'].str.contains('Pseudogene'),'Gene Symbol'].nunique()

    hypergeometric_df['method'] = 'Hypergeometric'
    hypergeometric_df['score'] = hypergeometric_df['Average -log(p-values)']   #use -log pvalue as score since higher is better 
    hypergeometric_df['n_sgrnas'] = hypergeometric_df['Number of perturbations'] #column that indicates # guides per gene
    #limit to genes for which the negative selection significance was provided
    hypergeometric_df['is_ascending'] = hypergeometric_df.apply(lambda row: np.abs(np.mean([float(x) for x in row['Individual ascending -log(p-values)'].split(';')])-row['Average -log(p-values)']) < 0.01,
                                    axis=1)
    hypergeometric_df = hypergeometric_df[hypergeometric_df['is_ascending']]
    hypergeometric_df['p_value'] = np.power(10,-1*hypergeometric_df['Average -log(p-values)'])
    #get False Discovery Rate 
    hypergeometric_df['rank'] = hypergeometric_df.index + 1
    hypergeometric_df['cumnum_controls'] = np.cumsum(hypergeometric_df['Gene Symbol'].str.contains('Pseudogene'))
    hypergeometric_df['FDR'] = hypergeometric_df['cumnum_controls']/num_pseudogene
    hypergeometric_df = hypergeometric_df[['Gene Symbol','p_value','n_sgrnas','method','FDR']]
    
    mageck_df['method'] = 'MAGeCK'
    mageck_df['score'] = -np.log10(mageck_df['neg|p-value'])
    mageck_df['n_sgrnas'] = mageck_df['num']
    mageck_df = mageck_df.rename(columns = {'id':'Gene Symbol','neg|p-value':'p_value'})
    mageck_df['rank'] = mageck_df.index + 1
    mageck_df['cumnum_controls'] = np.cumsum(mageck_df['Gene Symbol'].str.contains('Pseudogene'))
    mageck_df['FDR'] = mageck_df['cumnum_controls']/num_pseudogene
    mageck_df = mageck_df[['Gene Symbol','p_value','n_sgrnas','method','FDR']]

    zscore_df['method']='Z-Score'
    #Not calculating FDR for Z-score method because this is already done (in Jacquere Screen performance notebooks)
    zscore_df=zscore_df.rename(columns={"Target Gene Symbol":"Gene Symbol"})
    zscore_df = zscore_df[['Gene Symbol','p_value','n_sgrnas','method','FDR']]

    return pd.concat([zscore_df,hypergeometric_df,mageck_df])
```

```python
brunello_hitcallingresults=process_hitcalling_results(brunello_zscore_results,brunello_hypergeometric_results,brunello_mageck_results)
jacquere_hitcallingresults=process_hitcalling_results(jacquere_zscore_results,jacquere_hypergeometric_results,jacquere_mageck_results)
```

```python
methods = ['Z-Score','Hypergeometric','MAGeCK']
color_map = {'Mageck':'blue','Z-Score':'green','Hypergeometric':'orange'}
```

```python
#guide-level info for brunello, jacquere
screen_results=pd.read_csv("../../6. Jacquere Screen Performance/Data/jacquere_brunello_guideinfo.csv")
```

```python
jacquere['library'] = 'Jacquere'
brunello['library'] = 'Brunello'
combined_df = pd.concat([jacquere,brunello])
combined_df = combined_df[combined_df['Gene Symbol'] != '(ENSG000']
```

```python
combined_df['-log10FDR'] = np.where(combined_df['FDR'] > 0, -np.log10(combined_df['FDR']), -np.log10(1/num_pseudogene))
combined_df['-log10pvalue'] = -np.log10(combined_df['p-value'])                                   
max_finite_logpval = combined_df[np.isfinite(combined_df['-log10pvalue'])].groupby(['method','library']).agg(max_finite_log10pval = ('-log10pvalue',
                                                                                                                         'max')).reset_index()
```

```python
combined_df = combined_df.merge(max_finite_logpval, on = ['method','library'], how = 'inner')
combined_df
```

```python
combined_df['-log10pvalue'] = np.where(np.isfinite(combined_df['-log10pvalue']),combined_df['-log10pvalue'],combined_df['max_finite_log10pval'])
combined_df
```

```python
three_four_guide_df = combined_df[((combined_df['n_sgrnas'] == 3) & (combined_df['library'] == 'Jacquere')) |
                                  ((combined_df['n_sgrnas'] == 4) & (combined_df['library'] == 'Brunello'))]
plot_df = three_four_guide_df.pivot(index = ['Gene Symbol','method'], columns = 'library', values = '-log10pvalue').reset_index()
plot_df = plot_df.dropna().reset_index(drop=True)
essential_genes = pd.read_table('https://raw.githubusercontent.com/gpp-rnd/genesets/master/human/essential-genes-Hart2015.txt',
                                names=['gene']).gene
plot_df = plot_df[(plot_df['Gene Symbol'].isin(essential_genes))].reset_index(drop=True)

(fig,ax)=plt.subplots(1,len(methods),figsize=(20,5))
sns.despine()

for i in range(len(methods)):
    method = methods[i]
    subplot_df = plot_df[plot_df['method'] == method]
    gpp.point_densityplot(ax = ax[i], x = 'Brunello', y = 'Jacquere', data = subplot_df)
    ax[i].axline((0,0),slope=1,color='red',ls='dashed')
    ax[i].set_title(method)
    min_score = subplot_df[['Brunello','Jacquere']].min().min() - 1
    max_score = subplot_df[['Brunello','Jacquere']].max().max() + 1
    ax[i].set_xlim(min_score,max_score)
    ax[i].set_ylim(min_score,max_score)

    subplot_df['Brunello > Jacquere'] = subplot_df['Brunello'] > subplot_df['Jacquere']
    pct_brunello_higher = round(subplot_df['Brunello > Jacquere'].mean() * 100)

    ax[i].text(min_score + 0.7*(max_score-min_score),
               min_score + 0.1*(max_score-min_score),
               str(pct_brunello_higher) + '% genes score \nstronger in Brunello')

fig.suptitle('-log10(p-value) reported for essential genes')

gpp.savefig('../Figures/screen_analysis_methods_jacquere_vs_brunello.pdf',dpi=300,bbox_inches='tight')
```

```python
plot_df = three_four_guide_df[three_four_guide_df['Gene Symbol'].isin(essential_genes)].reset_index(drop=True)
(fig,ax)=plt.subplots(1,len(methods),figsize=(20,5))
sns.despine()

for i in range(len(methods)):
    method = methods[i]
    subplot_df = plot_df[plot_df['method'] == method]
    sns.kdeplot(ax = ax[i], x = '-log10pvalue', hue = 'library', data = subplot_df,
               cumulative = True, common_norm = False, common_grid = False)
    ax[i].set_title(method)
    ax[i].set_xlabel('-log10(p-value) cutoff')
    ax[i].set_ylabel('Proportion of essential genes that are insignificant')

gpp.savefig('../Figures/screen_analysis_methods_jacquere_vs_brunello_cumulative_density.pdf',dpi=300,bbox_inches='tight')
```
