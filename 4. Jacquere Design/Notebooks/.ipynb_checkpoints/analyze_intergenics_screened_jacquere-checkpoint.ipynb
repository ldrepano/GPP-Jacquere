```python
import pandas as pd
import matplotlib.pyplot as plt
import gpplot as gpp
```

```python
#CRISPick run that contains the intergenic guides screened with the Jacquere library 
jacquere_intergenics_screened=pd.read_table("../Data/scoring_intergenics_screened_jacquere-sgrna-designs.txt")["sgRNA Sequence"].tolist()

#CRISPick run that targets many intergenic sites by their coordinates to retrieve Aggregate CFD and RS3 score 
crispick_intergenics=pd.read_table("../Data/Negative Controls/jacquere-intergenics-sgrna-designs.txt",usecols=["sgRNA Sequence","On-Target Efficacy Score","Aggregate CFD Score"])
crispick_intergenics=crispick_intergenics[crispick_intergenics["sgRNA Sequence"].isin(jacquere_intergenics_screened)]
crispick_intergenics.to_csv("../Data/Negative Controls/jacquere_intergenics.csv")
```

```python
#retrieve gene targeting guides
jacquere=pd.read_csv("../Data/jacquere_assembled_crispick.csv",low_memory=False)
```

```python
print("Mean RS3 score of gene-targeting guides:", round(jacquere["On-Target Efficacy Score"].mean(),2))
print("Mean RS3 score of intergenic-targeting guides:", round(crispick_intergenics["On-Target Efficacy Score"].mean(),2))
print("\nMedian RS3 score of gene-targeting guides:", round(jacquere["On-Target Efficacy Score"].median(),2))
print("Median RS3 score of intergenic-targeting guides:", round(crispick_intergenics["On-Target Efficacy Score"].median(),2))
```

```python
(fig,ax)=plt.subplots(1,2,figsize=(16,5), sharex='col',layout="tight")
jacquere["Aggregate CFD Score"].hist(grid=False,bins=30,ax=ax[0])
crispick_intergenics["Aggregate CFD Score"].hist(grid=False,bins=30,ax=ax[1])

fig.suptitle("Aggregate CFD distribution of gene-targeting vs intergenic guides in Jacquere screen",fontsize=18)
ax[0].set_ylabel("# Gene-targeting Guides",fontsize=18)
ax[1].set_ylabel("# Intergenic Guides",fontsize=18)
gpp.savefig("../Figures/jacquere_screen_aggcfd_distribution_intergenics_targeting.pdf",bbox_inches="tight",dpi=600)
```
